## **I. Load a Video**

```python
import os
import tempfile
import cv2 as cv
from base64 import b64encode
from IPython.display import HTML, display

# Define the URL for the video
video_url = 'https://github.com/spmallick/learnopencv/raw/refs/heads/master/Human-Action-Recognition-Using-Detectron2-And-Lstm/sample_video.mp4'

# Create a VideoCapture object
cap = cv.VideoCapture(video_url)

# Check if video opened successfully
if not cap.isOpened():
  print(f"- [Error] Could not open video from the URL!")
else:
  # Get video properties for VideoWriter
  frame_width = int(cap.get(cv.CAP_PROP_FRAME_WIDTH))
  frame_height = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))
  fps = int(cap.get(cv.CAP_PROP_FPS))
  if fps == 0:
      fps = 30

  # Create temporary files for video output
  temp_output_path = tempfile.NamedTemporaryFile(suffix='.mp4', delete=False).name
  final_output_path = tempfile.NamedTemporaryFile(suffix='.mp4', delete=False).name

  # Define the codec and create VideoWriter object
  # 'mp4v' codec works well for temporary storage before ffmpeg conversion
  fourcc = cv.VideoWriter_fourcc(*'mp4v')
  out = cv.VideoWriter(temp_output_path, fourcc, fps, (frame_width, frame_height))

  while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Write the frame to the output video file
    out.write(frame)

  # Release video capture and writer objects
  cap.release()
  out.release()

  print("- Video processing complete. Converting for HTML playback...")

  # Use ffmpeg to convert the video to a web-friendly format (H.264)
  # -y overwrites output file without asking
  # -i specifies input file
  # -vcodec libx264 specifies the H.264 codec
  # -vf "format=yuv420p" ensures compatibility with various players
  # -loglevel quiet suppresses verbose output from ffmpeg
  os.system(f"ffmpeg -y -i {temp_output_path} -vcodec libx264 -vf \"format=yuv420p\" {final_output_path} -loglevel quiet")

  # Read the final video file as bytes
  mp4 = open(final_output_path, 'rb').read()

  # Encode to base64 for embedding in HTML
  data_url = "data:video/mp4;base64," + b64encode(mp4).decode()

  # Display the video using HTML
  display(HTML(f'''
      <video controls width="600">
          <source src="{data_url}" type="video/mp4">
      </video>
  '''))

  # Clean up temporary files
  os.remove(temp_output_path)
  os.remove(final_output_path)
```
